````python

import random
board = [[" " for _ in range(3)] for _ in range(3)]
def display_board():
    print("\n")
    for row in board:
        print(" | ".join(row))
        print("-" * 9)
def check_winner(symbol):
    for row in board:
        if row[0] == row[1] == row[2] == symbol:
            return True
    for col in range(3):
        if board[0][col] == board[1][col] == board[2][col] == symbol:
            return True
    if board[0][0] == board[1][1] == board[2][2] == symbol:
        return True
    if board[0][2] == board[1][1] == board[2][0] == symbol:
        return True
    return False
def is_draw():
    for row in board:
        if " " in row:
            return False
    return True
def human_move():
    while True:
        try:
            row = int(input("Enter row (0-2): "))
            col = int(input("Enter column (0-2): "))
            if board[row][col] == " ":
                board[row][col] = "X"
                break
            else:
                print("Cell already occupied.")
        except:
            print("Invalid input. Try again.")
def computer_move():
    empty_cells = []
    for i in range(3):
        for j in range(3):
            if board[i][j] == " ":
                empty_cells.append((i, j))
    row, col = random.choice(empty_cells)
    board[row][col] = "O"
print("TIC-TAC-TOE")
print("Human = X")
print("Computer = O")
display_board()
while True:
    human_move()
    display_board()
    if check_winner("X"):
        print("Human Wins!")
        break
    if is_draw():
        print("Game Draw!")
        break
    print("Computer's Turn...")
    computer_move()
    display_board()
    if check_winner("O"):
        print("Computer Wins!")
        break
    if is_draw():
        print("Game Draw!")
        break
````

### 8-Puzzle Problem Solver using Depth-First Search (DFS)

This code implements a solution for the 8-puzzle problem, a sliding puzzle that consists of a 3x3 grid with 8 numbered tiles and one blank space. The goal is to rearrange the tiles into a specific target configuration by repeatedly moving a tile into the blank space.

We will use Depth-First Search (DFS) to explore possible moves and find a path from an initial state to the goal state.

````python
import collections

# Define the goal state for the 8-puzzle
GOAL_STATE = ((1, 2, 3), (4, 5, 6), (7, 8, 0)) # 0 represents the blank space

def print_puzzle(state):
    """Prints the 8-puzzle state in a readable format."""
    for row in state:
        print(' '.join(map(str, row)).replace('0', ' '))
    print()

def find_blank(state):
    """Finds the row and column of the blank (0) tile."""
    for r in range(3):
        for c in range(3):
            if state[r][c] == 0:
                return r, c
    return -1, -1 # Should not happen in a valid puzzle

def generate_moves(state):
    """Generates all possible next states from the current state."""
    r_blank, c_blank = find_blank(state)
    moves = []
    # Possible directions: (dr, dc) for (row_change, col_change)
    directions = [(-1, 0), (1, 0), (0, -1), (0, 1)] # Up, Down, Left, Right

    for dr, dc in directions:
        new_r, new_c = r_blank + dr, c_blank + dc

        # Check if the new position is within the board boundaries
        if 0 <= new_r < 3 and 0 <= new_c < 3:
            # Create a new state by swapping the blank with the tile at (new_r, new_c)
            new_state_list = [list(row) for row in state]

            # Swap
            new_state_list[r_blank][c_blank] = new_state_list[new_r][new_c]
            new_state_list[new_r][new_c] = 0

            moves.append(tuple(tuple(row) for row in new_state_list))
    return moves

def dfs_solver(initial_state):
    """Solves the 8-puzzle using Depth-First Search (DFS)."""
    # Stack for DFS: stores (current_state, path_to_current_state)
    stack = collections.deque([(initial_state, [initial_state])])

    # Set to store visited states to avoid cycles
    visited = {initial_state}

    while stack:
        current_state, path = stack.pop()

        if current_state == GOAL_STATE:
            return path # Goal reached!

        for next_state in generate_moves(current_state):
            if next_state not in visited:
                visited.add(next_state)
                # Push new state and extended path onto the stack
                stack.append((next_state, path + [next_state]))

    return None # No solution found
````

#### Example Usage:

Let's try to solve an 8-puzzle from an initial configuration to the goal state.

````python
# Example initial state (ensure it's solvable for DFS to find a path)
# A common solvable example:
initial_puzzle_state = (
    (1, 2, 3),
    (0, 4, 6),
    (7, 5, 8)
)

print("Initial Puzzle State:")
print_puzzle(initial_puzzle_state)

print("Goal State:")
print_puzzle(GOAL_STATE)

print("Solving...")
solution_path = dfs_solver(initial_puzzle_state)

if solution_path:
    print(f"Solution Found in {len(solution_path) - 1} moves!")
    for i, state in enumerate(solution_path):
        print(f"Step {i}:")
        print_puzzle(state)
else:
    print("No solution found for this initial state (or it's unsolvable by DFS without deeper exploration constraints).")

# Another example: A slightly harder one
initial_puzzle_state_2 = (
    (1, 2, 3),
    (4, 5, 6),
    (0, 7, 8)
)

print("\nInitial Puzzle State 2:")
print_puzzle(initial_puzzle_state_2)

print("Solving...")
solution_path_2 = dfs_solver(initial_puzzle_state_2)

if solution_path_2:
    print(f"Solution Found in {len(solution_path_2) - 1} moves!")
    for i, state in enumerate(solution_path_2):
        print(f"Step {i}:")
        print_puzzle(state)
else:
    print("No solution found for this initial state (or it's unsolvable by DFS without deeper exploration constraints).")
````
