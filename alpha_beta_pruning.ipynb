```python
def alpha_beta_pruning(node, depth, alpha, beta, is_max, tree, path=[]):
    # A node is a leaf if it does not point to a list of children in our game tree
    if node not in tree or not isinstance(tree[node], list):
        return tree[node], path + [node]

    best_path = []
    if is_max:
        max_eval = float('-inf')
        for child in tree[node]:
            evaluation, child_path = alpha_beta_pruning(child, depth + 1, alpha, beta, False, tree, path + [node])
            if evaluation > max_eval:
                max_eval = evaluation
                best_path = child_path
            alpha = max(alpha, evaluation)
            if beta <= alpha:
                print(f"Pruned at node: {node} (Max Mode) during child: {child}")
                break  # Beta cutoff
        return max_eval, best_path
    else:
        min_eval = float('inf')
        for child in tree[node]:
            evaluation, child_path = alpha_beta_pruning(child, depth + 1, alpha, beta, True, tree, path + [node])
            if evaluation < min_eval:
                min_eval = evaluation
                best_path = child_path
            beta = min(beta, evaluation)
            if beta <= alpha:
                print(f"Pruned at node: {node} (Min Mode) during child: {child}")
                break  # Alpha cutoff
        return min_eval, best_path

# Correct representation: non-leaf nodes map to lists of children;
# leaf nodes map to numeric values.
game_tree = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': ['H', 'I'],
    'E': ['J', 'K'],
    'F': ['L', 'M'],
    'G': ['N', 'O'],
    'H': 3, 'I': 5, 'J': 6, 'K': 9,
    'L': 1, 'M': 2, 'N': 0, 'O': -1
}

# Run Alpha-Beta Pruning
root_node = 'A'
optimal_value, optimal_path = alpha_beta_pruning(root_node, depth=0, alpha=float('-inf'), beta=float('inf'), is_max=True, tree=game_tree)

print(f"\nRoot Node value (MAX): {optimal_value}")
print(f"Optimal path: {' -> '.join(optimal_path)}")
```
