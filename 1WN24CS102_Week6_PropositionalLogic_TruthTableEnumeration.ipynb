### Propositional Logic Truth-Table Enumeration Implementation

```python
import pandas as pd

# Define the basic operators
def implication(p, q):
    return not p or q

def biconditional(p, q):
    return p == q

# Generate all possible models (combinations of truth values for P, Q, R)
symbols = ['P', 'Q', 'R']
models = []
for p_val in [True, False]:
    for q_val in [True, False]:
        for r_val in [True, False]:
            models.append({'P': p_val, 'Q': q_val, 'R': r_val})

# Evaluate KB sentences and queries for each model
results = []
for model in models:
    p, q, r = model['P'], model['Q'], model['R']

    # KB Sentences
    kb1 = biconditional(p, q)             # P ↔ Q
    kb2 = q or r                           # Q ∨ R
    kb3 = implication(not p, r)            # P → R
    kb_all = kb1 and kb2 and kb3           # Entire KB holds

    # Queries
    q_ii = q                               # Q
    q_iii = implication(p, q)              # P → Q
    q_iv = implication(r, q)               # R → Q

    results.append({
        'P': p, 'Q': q, 'R': r,
        'P ↔ Q (KB1)': kb1,
        'Q ∨ R (KB2)': kb2,
        'P → R (KB3)': kb3,
        'KB Holds': kb_all,
        'Query ii (Q)': q_ii,
        'Query iii (P → Q)': q_iii,
        'Query iv (R → Q)': q_iv
    })

df_truth_table = pd.DataFrame(results)
display(df_truth_table)
```

```python
# Identify Models of the KB (where the KB is True)
kb_models = df_truth_table[df_truth_table['KB Holds'] == True]
print("Models where KB is True (Valid Models):")
display(kb_models[['P', 'Q', 'R', 'KB Holds']])

# Check Entailments
# KB entails ̑ iff in every model where KB is True, ̑ is also True.
for query_name, col_name in [
    ("ii) Q", "Query ii (Q)"),
    ("iii) P → Q", "Query iii (P → Q)"),
    ("iv) R → Q", "Query iv (R → Q)")
]:
    # Find if there are any models where KB is True but Query is False
    counterexamples = kb_models[kb_models[col_name] == False]
    entails = len(counterexamples) == 0
    print(f"\nDoes KB entail {query_name}? {entails}")
    if not entails:
        print("Counterexample model found:")
        display(counterexamples[['P', 'Q', 'R', col_name]])
```
