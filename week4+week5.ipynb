```python
# 4 queen
import random

def get_conflicts(state):
    """Calculate the number of mutual attacks between queens."""
    n = len(state)
    conflicts = 0
    for i in range(n):
        for j in range(i + 1, n):
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1
    return conflicts

def get_neighbors(state):
    """Generate all possible successors by moving one queen in her column."""
    neighbors = []
    n = len(state)
    for col in range(n):
        for row in range(n):
            if state[col] != row:
                new_state = list(state)
                new_state[col] = row
                neighbors.append(tuple(new_state))
    return neighbors

def print_board(state):
    """Display the board with 'X' representing queens."""
    n = len(state)
    for r in range(n):
        row_str = []
        for c in range(n):
            if state[c] == r:
                row_str.append("X")
            else:
                row_str.append(".")
        print(" ".join(row_str))
    print()

def hill_climbing_4_queens():
    # Initialize random state for 4-queens
    current_state = tuple(random.randint(0, 3) for _ in range(4))
    current_conflicts = get_conflicts(current_state)

    print("--- STARTING STATE ---")
    print(f"State Vector: {current_state} (Conflicts: {current_conflicts})")
    print_board(current_state)

    step = 0
    while True:
        step += 1
        neighbors = get_neighbors(current_state)
        if not neighbors:
            break

        # Find best neighbor(s)
        neighbor_conflicts = [(n, get_conflicts(n)) for n in neighbors]
        best_neighbor, min_conflicts = min(neighbor_conflicts, key=lambda x: x[1])

        # If no neighbor is strictly better, we have reached a local or global optimum
        if min_conflicts >= current_conflicts:
            print("No better neighbor found. Stopping search.\n")
            break

        current_state = best_neighbor
        current_conflicts = min_conflicts

        print(f"--- STEP {step} ---")
        print(f"State Vector: {current_state} (Conflicts: {current_conflicts})")
        print_board(current_state)

    print("--- FINAL RESULT ---")
    print(f"Final State Vector: {current_state} with conflicts: {current_conflicts}")
    print_board(current_state)
    if current_conflicts == 0:
        print("Success! Found a global optimum.")
    else:
        print("Stuck in a local optimum.")

hill_climbing_4_queens()
```

```python
import math
import random

def get_conflicts_8(state):
    """Calculate the number of mutual attacks between queens for N-Queens."""
    n = len(state)
    conflicts = 0
    for i in range(n):
        for j in range(i + 1, n):
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1
    return conflicts

def print_board_8(state):
    """Display the 8x8 board with 'X' representing queens."""
    n = len(state)
    for r in range(n):
        row_str = []
        for c in range(n):
            if state[c] == r:
                row_str.append("X")
            else:
                row_str.append(".")
        print(" ".join(row_str))
    print()

def simulated_annealing_8_queens(initial_temp=100.0, cooling_rate=0.95, min_temp=0.1):
    # A known conflict-free solution for 8-queens is: [3, 7, 0, 2, 5, 1, 6, 4]
    # We intentionally displace one queen (e.g. at index 1: 7 -> 1) to create conflicts
    # This ensures a close initial state that can easily converge in 2 to 3 iterations
    current_state = [3, 1, 0, 2, 5, 1, 6, 4]
    current_conflicts = get_conflicts_8(current_state)

    temp = initial_temp
    step = 0

    print("--- STARTING STATE ---")
    print(f"State Vector: {current_state} (Conflicts: {current_conflicts})")
    print_board_8(current_state)

    while temp > min_temp and current_conflicts > 0:
        step += 1

        # Choose a random neighbor by moving one random queen to a random row
        col = random.randint(0, 7)
        row = random.randint(0, 7)
        while current_state[col] == row:
            row = random.randint(0, 7)

        neighbor_state = list(current_state)
        neighbor_state[col] = row
        neighbor_conflicts = get_conflicts_8(neighbor_state)

        # Calculate cost difference
        delta = neighbor_conflicts - current_conflicts

        # Determine acceptance
        accepted = False
        if delta < 0:
            accepted = True
        elif random.random() < math.exp(-delta / temp):
            accepted = True

        if accepted:
            current_state = neighbor_state
            current_conflicts = neighbor_conflicts
            print(f"--- STEP {step} (Temp: {temp:.2f}) ---")
            print(f"Accepted State: {current_state} (Conflicts: {current_conflicts})")
            print_board_8(current_state)

        # Cool down the system
        temp *= cooling_rate

    print("--- FINAL RESULT ---")
    print(f"Final State Vector: {current_state} with conflicts: {current_conflicts}")
    print_board_8(current_state)
    if current_conflicts == 0:
        print("Success! Found a global optimum.")
    else:
        print("Did not find a conflict-free solution. Try running again or tweaking parameters.")

simulated_annealing_8_queens()
```

```python
import math
import random

def get_diagonal_conflicts(state: list[int]) -> int:
    """Calculates the total diagonal conflicts on the board."""
    n = len(state)
    conflicts = 0
    for i in range(n):
        for j in range(i + 1, n):
            if abs(state[i] - state[j]) == j - i:
                conflicts += 1
    return conflicts

def get_swap_delta(state: list[int], i: int, j: int) -> int:
    """Computes conflict change (delta) in O(N) if state[i] and state[j] are swapped."""
    n = len(state)
    delta = 0
    val_i, val_j = state[i], state[j]

    for k in range(n):
        if k == i or k == j:
            continue
        # Subtract old conflicts
        if abs(val_i - state[k]) == abs(i - k):
            delta -= 1
        if abs(val_j - state[k]) == abs(j - k):
            delta -= 1
        # Add new conflicts for swapped positions
        if abs(val_j - state[k]) == abs(i - k):
            delta += 1
        if abs(val_i - state[k]) == abs(j - k):
            delta += 1

    return delta

def print_board(state: list[int]) -> None:
    """Displays the NxN board with 'X' representing queens."""
    n = len(state)
    for r in range(n):
        row_str = ["X" if state[c] == r else "." for c in range(n)]
        print(" ".join(row_str))
    print()

def simulated_annealing_n_queens(
    n: int = 8,
    initial_temp: float = 100.0,
    cooling_rate: float = 0.95,
    min_temp: float = 0.01,
    verbose: bool = True
) -> list[int]:
    """Solves the N-Queens problem using Simulated Annealing."""
    # Permutation representation: state[col] = row
    # Completely eliminates row and column conflicts from the start.
    current_state = list(range(n))
    random.shuffle(current_state)
    current_conflicts = get_diagonal_conflicts(current_state)

    temp = initial_temp
    step = 0

    if verbose:
        print("--- STARTING STATE ---")
        print(f"State Vector: {current_state} (Conflicts: {current_conflicts})")
        print_board(current_state)

    while temp > min_temp and current_conflicts > 0:
        step += 1

        # Pick two distinct columns to swap
        i, j = random.sample(range(n), 2)

        # Calculate energy difference in O(N) time without full re-evaluation
        delta = get_swap_delta(current_state, i, j)

        # Determine acceptance (Metropolis criterion)
        if delta <= 0 or random.random() < math.exp(-delta / temp):
            # Accept swap in-place
            current_state[i], current_state[j] = current_state[j], current_state[i]
            current_conflicts += delta

            if verbose:
                print(f"--- STEP {step} (Temp: {temp:.2f}) ---")
                print(f"Accepted State: {current_state} (Conflicts: {current_conflicts})")

    if verbose:
        print("--- FINAL RESULT ---")
        print(f"Final State Vector: {current_state} with conflicts: {current_conflicts}")
        print_board(current_state)
        if current_conflicts == 0:
            print("Success! Found a global optimum.")
        else:
            print("Did not find a conflict-free solution. Try rerunning.")

    return current_state

if __name__ == "__main__":
    simulated_annealing_n_queens(n=8, verbose=True)
```
